## 2 · It is Monday, 9:00

Your CEO sends one message to the company analytics channel:

> **"Why did revenue drop 12% last week?"**

Take thirty seconds. In a company you have worked at, interned at, or can imagine —
**how long until there is a trustworthy answer, and who produces it?**

<br>

Typical honest answers: *two days*, *a week*, *nobody knows*, *three people answer and
disagree*. Rarely: *ten minutes*.

Now the part that makes this a 2026 course rather than a 2006 one.

### The twist

An LLM will write the SQL for that question in about ten seconds. It will be syntactically
perfect. It may even be elegant.

**And the answer still takes two days.** Why?

Because the SQL was never the hard part (even if it can be but tricky ... we'll talk about it when we'll have some spare time). The hard part is everything the SQL silently
assumes:

* *which* table — there are four with "orders" in the name
* whose definition of **revenue** — with or without VAT? shipping? cancelled orders? refunds?
* at what **grain** — per order, per line, per customer, per day?
* is last week's data even **loaded** yet, and complete?
* and when two people answer, **why do they disagree**, and which one do we act on?

> The bottleneck in analytics was never writing queries. It is agreeing on what the
> question means, and being able to trust the answer. LLMs made the easy part free
> and left the hard part exactly where it was — arguably made it *worse*, because now
> wrong answers arrive much faster and sound much more confident.

**Everything in these thirteen weeks is an attack on that gap.**

### The same question, asked twice

I put our raw order data in front of a very good language model and asked it what our
revenue was. Then I asked again, changing nothing but the phrasing:

> **"What was our revenue in 2025?"**
> `26,243,024 CZK`
>
> **"How much did we earn last year?"**
> `24,885,652 CZK`

Both answers are confident. Neither is flagged as uncertain. Neither is *wrong*, exactly —
the first counts completed orders, the second also subtracts refunds. Nobody told the model
which one we meant, because nobody in the company had ever written it down.

A difference of 1.4 million CZK, produced twice in ten seconds, by a tool that cannot tell
you it just made a judgement call on your behalf.

A few minutes later you produced six different numbers of your own, without any help from
an LLM — see §5.

## 3 · So what is Business Intelligence?

Wikipedia will tell you about [Business Intelligence](https://en.wikipedia.org/wiki/Business_intelligence),
[decision support systems](https://en.wikipedia.org/wiki/Decision_support_system) and
[OLAP](https://en.wikipedia.org/wiki/Online_analytical_processing). Useful, but let me
give you four claims instead. If you remember nothing else from today, remember these.

#### Claim 1 — Separate the systems that *run* the business from the systems that *understand* it

The e-shop database exists to take an order in 50 milliseconds without losing it.
Asking it "what were our best-selling categories by region last quarter" is both slow
and dangerous — you are competing for resources with actual customers.

Different job → different data structures → different system. We spend week 2 on exactly
this distinction.

#### Claim 2 — A metric has *one* definition, and that definition is the real deliverable

Not the dashboard. Not the chart. The agreement about what "revenue", "active customer"
and "churn" mean, written down once, in executable form, and inherited by every report
in the company.

Most BI projects fail here, not in the technology.

#### Claim 3 — History is a first-class citizen

Production systems overwrite. A customer changes their address and the old one is gone.
Analytics **remembers** — because "how many customers moved from the Casual to the
Competitive segment last year" is unanswerable if the database only knows *now*.

This turns out to be surprisingly hard, and has a name:
[slowly changing dimensions](https://en.wikipedia.org/wiki/Slowly_changing_dimension).

#### Claim 4 — The output is a changed decision, not a produced chart

A dashboard nobody acts on is a cost, not an asset. Throughout this course I will keep
asking you the same annoying question about every chart you build:

> **Who is going to do what differently because of this?**

If there is no answer, you built decoration.

## 4 · Your turn

Ten minutes, on your own:
[`handouts/01_what_is_revenue.pdf`](../handouts/01_what_is_revenue.pdf).

Seven order lines, one refund, one question:

> **What was our revenue?**

| order | line | product | qty | unit price (net) | discount | shipping | VAT | status |
|---|---|---|---:|---:|---:|---:|---:|---|
| O-1 | 1 | Running shoes | 2 | 1 000 | 0 | 100 | 21% | completed |
| O-1 | 2 | GPS watch | 1 | 3 000 | 500 | 100 | 21% | completed |
| O-2 | 1 | Wetsuit | 1 | 5 000 | 0 | 0 | 21% | completed |
| O-3 | 1 | Trail shoes | 3 | 1 000 | 300 | 100 | 21% | **cancelled** |
| O-4 | 1 | Bike helmet | 1 | 2 000 | 0 | 100 | 21% | completed |
| O-5 | 1 | Goggles | 2 | 1 500 | 0 | 150 | 21% | **pending** |
| O-6 | 1 | GPS watch | 1 | 4 000 | 0 | 100 | 21% | completed |

*refunds:* R-1 refunds order O-4 in full, 2 000 net, three weeks later.
*Shipping is charged once per order and repeated on every line. Prices exclude VAT.*

Two things to write down: **the number, and the rule you used to get it.** The rule matters
more.

## 5 · How it went

Eight of you answered. Every response, sorted by the number, with the rule lightly shortened:

| answer (CZK) | counted | excluded |
|---:|---|---|
| **9 850** | (shipping per line + unit × qty − discount) × (1 − VAT) | pending, cancelled; refund subtracted |
| **11 929** | — | — |
| **11 929** | qty × unit price − shipping, then tax | pending, cancelled, refund |
| **13 400** | all of the sold items | VAT, the refunded item and its shipping |
| **13 500** | (unit price − discount) × qty, no shipping | trail shoes, goggles, bike helmet |
| **13 500** | O-1, O-2, O-4, O-6 = 15 500, minus refund R-1 | VAT, shipping, cancelled O-3, pending O-5 |
| **18 210** | (completed − discounts + shipping per order) × VAT | cancelled and pending |
| **19 240** | completed | cancelled and pending |

![Every response, against the six rules](../files/week01_responses.png)

**Six different numbers from eight people, between 9 850 and 19 240.** The largest answer
is 95% bigger than the smallest — almost double. That is a *wider* spread than the six
clean rules on the handout can produce (13 500 to 22 000, 63%), because on top of choosing
a rule, the arithmetic went its own way.

### What happened

**One clean rule, chosen twice.** 13 500 is rule D: completed orders, minus discounts, net
of the refund, no shipping, no VAT. Nobody else landed on any of the six rules exactly.

| | rule | answer |
|---|---|---:|
| A | everything in the table | 22 000 |
| B | ...minus discounts | 21 200 |
| C | ...completed orders only | 15 500 |
| D | ...net of the refund | **13 500** |
| E | ...plus shipping, which we do charge for | 15 800 |
| F | ...as finance reports it, with VAT | 18 755 |

**A seventh rule nobody planned — and a good one.** 13 400 is rule D minus another 100 CZK:
the shipping on the refunded order, on the reasoning that the company paid to deliver goods
it then refunded. Perfectly defensible.

**The best answers were questions.** The same response asked: *is VAT counted in revenue?
Is this "net revenue"? Who paid the return shipping — the company or the customer? Was the
wetsuit picked up in the store, or shipped for free?* The data cannot answer any of these.
The business has to — once, for everybody. That is what a semantic layer is for.

**Shipping went three ways:** ignored (13 500), added (9 850, 18 210, 19 240), and subtracted
as a cost (11 929). All three are positions a real company might take.

**VAT went both ways:** added (18 210, 19 240) and taken out (9 850, 11 929) — but the
prices were already net, so there was no VAT in them to take out. A related trap worth
knowing: taking 21% off is not the same as undoing a 21% increase. 100 × 1.21 = 121, but
121 × 0.79 = 95.59.

**The fan-out turned up, as promised.** 19 240 is, give or take rounding, 15 900 × 1.21:
the shipping column summed straight down the page, so order O-1 paid for its delivery
twice — and then VAT on top. 11 929 subtracts the same inflated 400 instead of 300, and the
rule behind 9 850 names it outright: *shipping per line*.

**Only three of the eight numbers can be reproduced from the rule written beside them.**
The rule behind 18 210 — completed orders, minus discounts, plus shipping per order, times
VAT — actually gives 19 118. The rule behind the second 11 929 says the refund was excluded;
its arithmetic did not exclude it. Neither slip would ever have been noticed without the
rule written next to the number — and the first 11 929 came with no rule at all, which means
it can be neither checked nor reused.

**A rule can be wrong and still be right.** *(unit price − discount) × qty* subtracts the
discount once per unit. On this data that makes no difference — the only discounted
completed line has a quantity of 1 — so it produced the correct 13 500. On tomorrow's data
it will not. A rule that passes today and fails tomorrow is the reason data tests exist.

None of this is a criticism. It is what happens to everyone, every time — which is the whole
reason the rest of this course exists.

> **Same question. Same seven rows. Same room.** The disagreement was never about the data.
> It was about what the question *meant*, and about arithmetic nobody could check. Next week
> SQL fixes the second problem. The rest of the semester is about the first.
